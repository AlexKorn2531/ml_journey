## Урок №1. Векторы и операции над ними

### 1) Что такое вектор
**Вектор** — это упорядоченный набор чисел $(x_1, x_2, \dots, x_n)$, представляющий точку или направленный отрезок в $n$-мерном пространстве. В ML вектор обычно записывается как вектор-столбец:
$$x = \begin{pmatrix} x_1 \\ x_2 \\ \dots \\ x_n \end{pmatrix}$$

### 2) Свойства векторов
* **Сложение:** $x + y = (x_1 + y_1, x_2 + y_2, \dots, x_n + y_n)$. Геометрически — правило параллелограмма.
* **Умножение на скаляр:** $\alpha x = (\alpha x_1, \alpha x_2, \dots, \alpha x_n)$. Изменяет длину вектора (и направление, если $\alpha < 0$).
* **Коммутативность и ассоциативность:** $x + y = y + x$ и $(x + y) + z = x + (y + z)$.

### 3) Скалярное произведение
**Скалярное произведение (Dot Product)** двух векторов — это число, равное сумме произведений их соответствующих координат или произведению их длин на косинус угла между ними:
$$\langle x, y \rangle = x^T y = \sum_{i=1}^n x_i y_i = \|x\| \cdot \|y\| \cdot \cos(\theta)$$
Если $\langle x, y \rangle = 0$, векторы **ортогональны** (перпендикулярны).

### 4) Норма вектора: L2, L1, L_inf
**Норма** — функция, задающая длину вектора (метрику расстояния от начала координат).
* **Евклидова норма ($L_2$):** Расстояние по прямой. На ней базируется MSE и $L_2$-регуляризация.
  $$\|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2}$$
* **Манхэттенская длина ($L_1$):** Сумма абсолютных смещений по осям («сетка улиц»). Используется в MAE и $L_1$-регуляризации.
  $$\|x\|_1 = \sum_{i=1}^n |x_i|$$
* **Норма Чебышева ($L_\infty$):** Максимальное абсолютное значение среди всех координат вектора.
  $$\|x\|_\infty = \max_{i} |x_i|$$

### 5) Косинус угла между векторами
Показывает геометрическую направленность векторов друг относительно друга без учета их масштаба. Находится из формулы скалярного произведения:
$$\cos(\theta) = \frac{x^T y}{\|x\|_2 \cdot \|y\|_2}$$

### 6) Косинусное расстояние, косинусная близость
* **Косинусная близость (Cosine Similarity):** Равна $\cos(\theta)$. Меняется от $-1$ (противоположные векторы) до $1$ (сонаправленные). Используется в NLP и рекомендательных системах.
* **Косинусное расстояние (Cosine Distance):** Метрика различия векторов. 
  $$\text{Distance} = 1 - \text{Similarity} = 1 - \cos(\theta)$$

### 7) Нормировка вектора
Процесс приведения длины вектора к единице без изменения его направления. Для этого каждая координата вектора делится на его евклидову норму:
$$\hat{x} = \frac{x}{\|x\|_2}, \quad \text{при этом } \|\hat{x}\|_2 = 1$$

### 8) Проекция вектора на вектор
* **Скалярная проекция** вектора $x$ на вектор $y$ — это длина тени, которую отбрасывает $x$ на прямую $y$: $\text{proj}_y x = \|x\|\cos(\theta) = \frac{x^T y}{\|y\|}$.
* **Векторная проекция** — это сам вектор этой тени:
$$\vec{\text{proj}}_y x = \left( \frac{x^T y}{\|y\|_2^2} \right) y$$

## Урок №2. Матрицы и линейные преобразования

### 1) Матрица
**Матрица** — это прямоугольная таблица чисел размером $m \times n$ ($m$ строк, $n$ столбцов). В ML матрица чаще всего представляет датасет, где строки — это объекты (сэмплы), а столбцы — признаки (фичи).

### 2) Транспонирование матриц
Операция отражения матрицы относительно её главной диагонали (строки становятся столбцами): $(A^T)_{ij} = A_{ji}$.
**Свойства:**
1. $(A^T)^T = A$
2. $(A + B)^T = A^T + B^T$
3. $(\alpha A)^T = \alpha A^T$
4. $(AB)^T = B^T A^T$ *(порядок матриц меняется!)*

### 3) Умножение матрицы на вектор
Операция вида $y = Ax$, где результатом является новый вектор. Каждый элемент $y_i$ — это скалярное произведение $i$-й строки матрицы $A$ на вектор-столбец $x$. Геометрически это означает **линейное преобразование** вектора $x$ (его поворот, растяжение или проекцию в другое пространство).

### 4) Матрица поворота
Специальная ортогональная матрица, которая при умножении на вектор поворачивает его на угол $\alpha$ вокруг начала координат против часовой стрелки без изменения длины:
$$R = \begin{pmatrix} \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha \end{pmatrix}$$

### 5) Умножение матриц
Операция $C = AB$ возможна, только если количество столбцов $A$ равно количеству строк $B$. Элемент $c_{ij}$ равен скалярному произведению $i$-й строки матрицы $A$ на $j$-й столбец матрицы $B$.
* **Важно:** Операция некоммутативна ($AB \neq BA$).

### 6) Единичная и обратная матрица
* **Единичная матрица ($I$ или $E$):** Квадратная матрица с единицами на главной диагонали и нулями в остальных ячейках. Выполняет роль «единицы» в матричном мире: $AI = IA = A$.
* **Обратная матрица ($A^{-1}$):** Квадратная матрица, удовлетворяющая условию $A \cdot A^{-1} = A^{-1} \cdot A = I$. Существует только для квадратных невырожденных матриц ($\det A \neq 0$).
* **Свойства:** $(A^{-1})^{-1} = A$, $(AB)^{-1} = B^{-1}A^{-1}$, $(A^T)^{-1} = (A^{-1})^T$.

### 7) Линейная регрессия: МНК, нормальное уравнение, регуляризация
* **Метод наименьших квадратов (МНК):** Минимизирует сумму квадратов ошибок между предсказаниями и реальными ответами: $\|Xw - y\|^2 \rightarrow \min_w$.
* **Формула нормального уравнения (аналитическое решение МНК):**
  $$w = (X^T X)^{-1} X^T y$$
* **Регуляризация:** Борьба с переобучением путем штрафа за слишком большие веса $w$:
  * **Ridge ($L_2$-регуляризация):** Добавляет штраф $\lambda \|w\|_2^2$. Решение: $w = (X^T X + \lambda I)^{-1} X^T y$ (избавляет от мультиколлинеарности).
  * **Lasso ($L_1$-регуляризация):** Добавляет штраф $\lambda \|w\|_1$. Зануляет неважные веса, производя отбор признаков (feature selection).

### 8) Линейная комбинация, линейная независимость
* **Линейная комбинация:** Вектор вида $v = c_1 v_1 + c_2 v_2 + \dots + c_k v_k$, где $c_i$ — скаляры.
* **Линейная независимость:** Набор векторов независим, если их линейная комбинация равна нулю только при всех нулевых коэффициентах $c_i = 0$. Если хотя бы один вектор можно выразить через другие, набор называется **линейно зависимым**.

### 9) Ранг и вырожденность матриц
* **Ранг матрицы ($\text{rank}(A)$):** Максимальное количество линейно независимых строк или столбцов.
* **Вырожденная (сингулярная) матрица:** Квадратная матрица, у которой строки или столбцы линейно зависимы. Её определитель (детерминант) равен нулю ($\det A = 0$), и для неё **не существует** обратной матрицы $A^{-1}$.

### 10) NumPy Broadcasting (Трансляция)
Механизм в NumPy, позволяющий проводить арифметические операции над массивами разных (но совместимых) размерностей. Меньший массив виртуально «дублируется» (растягивается) вдоль осей, чтобы соответствовать большему массиву, без реального копирования данных в памяти.

### 11) Аффинные преобразования
Преобразования, которые сохраняют параллельность прямых, но могут сдвигать начало координат. Состоят из линейного преобразования (умножение на матрицу $A$) и параллельного переноса (векторного сдвига $b$):
$$f(x) = Ax + b$$
В нейросетях любой полносвязный слой (Dense) вычисляет именно аффинное преобразование: $weights \cdot x + bias$.

### 12) Псевдообратная матрица, SVD, ортогональная матрица
* **Ортогональная матрица:** Квадратная матрица, у которой $A^T A = A A^T = I$. Её строки и столбцы — взаимно ортогональные единичные векторы. Обладает свойством $A^{-1} = A^T$.
* **Псевдообратная матрица (Мура-Пенроуза, $A^+$):** Обобщение обратной матрицы на случай прямоугольных или вырожденных матриц. Вычисляется через SVD: $A^+ = V \Sigma^+ U^T$. Позволяет решать системы уравнений, даже если решений нет или их бесконечно много.
* **SVD (Сингулярное разложение):** Разложение *любой* матрицы $A$ размера $m \times n$ на произведение трех матриц: $A = U \Sigma V^T$, где $U, V$ — ортогональные матрицы, а $\Sigma$ — диагональная матрица с сингулярными числами.

### 13) Метод Гаусса (прямой и обратный ход)
Алгоритм решения систем линейных уравнений (СЛАУ) путем приведения расширенной матрицы к ступенчатому виду с помощью элементарных строк:
* **Прямой ход:** Обнуление элементов под главной диагональю сверху вниз (превращение матрицы в верхнетреугольную).
* **Обратный ход:** Вычисление значений переменных снизу вверх, начиная с последней строки.

## Урок №3. Собственные числа и SVD

### 1) Собственные векторы и числа
Если при умножении квадратной матрицы $A$ на ненулевой вектор $v$ этот вектор лишь масштабируется (не меняет своего направления в пространстве), то $v$ называется **собственным вектором**, а масштаб $\lambda$ — **собственным числом (значением)**:
$$A v = \lambda v$$
Для поиска $\lambda$ решается характеристическое уравнение: $\det(A - \lambda I) = 0$.

### 2) Метод главных компонент PCA
Алгоритм снижения размерности данных с минимальной потерей информации.
1. Данные центрируются (вычитается среднее по колонкам).
2. Считается матрица ковариации признаков $C = \frac{1}{m} X^T X$.
3. Находятся собственные векторы матрицы ковариации.
4. Векторы с наибольшими собственными числами $\lambda$ объявляются **главными компонентами** (они указывают направления максимальной дисперсии данных). Данные проецируются на них.

### 3) Спектральное разложение квадратной вещественной симметричной матрицы
Любую квадратную симметричную матрицу $A$ ($A = A^T$) можно представить в виде:
$$A = Q \Lambda Q^T$$
Где $Q$ — ортогональная матрица, составленная из собственных векторов, а $\Lambda$ — диагональная матрица, состоящая из собственных чисел $\lambda_i$. В геометрии это означает переход к базису, где преобразование является простым растяжением вдоль осей.

### 4) LoRA - Low-Rank Adaptation
Метод эффективной дообучаемости больших языковых моделей (LLM). Вместо обновления огромной матрицы весов $W_0$ размера $d \times d$ (что дорого), изменение весов $\Delta W$ аппроксимируется произведением двух низкоранговых матриц низкого ранга $r$ ($r \ll d$):
$$W = W_0 + \Delta W = W_0 + B \cdot A$$
Где матрица $A$ имеет размер $r \times d$, а $B$ — $d \times r$. Это сокращает количество обучаемых параметров в тысячи раз.

## Урок №4. Производные и градиенты

### 1) Определение производной
Производная функции $f(x)$ в точке — это предел отношения приращения функции к приращению аргумента, когда приращение аргумента стремится к нулю:
$$f'(x) = \lim_{\Delta x \to 0} \frac{f(x + \Delta x) - f(x)}{\Delta x}$$
Геометрически — это тангенс угла наклона касательной к графику функции в данной точке, отражающий мгновенную скорость изменения функции.

### 2) Таблица производных основных функций
* Константа: $(c)' = 0$
* Степенная функция: $(x^n)' = n x^{n-1}$
* Экспонента: $(e^x)' = e^x$
* Натуральный логарифм: $(\ln x)' = \frac{1}{x}$
* Тригонометрические функции: $(\sin x)' = \cos x$, $(\cos x)' = -\sin x$

### 3) Функции активации
Вводят нелинейность в нейросети, позволяя им аппроксимировать сложные зависимости:
* **Сигмоида (Sigmoid):** Сжимает числа в диапазон $(0, 1)$. Применяется в логистической регрессии и бинарной классификации. 
  $$\sigma(x) = \frac{1}{1 + e^{-x}}, \quad \text{производная: } \sigma'(x) = \sigma(x)(1 - \sigma(x))$$
* **ReLU (Rectified Linear Unit):** Срезает отрицательные значения, защищая от затухания градиентов глубокие сети. 
  $$f(x) = \max(0, x), \quad \text{производная: } 1 \text{ при } x > 0 \text{ и } 0 \text{ при } x < 0$$
* **Softmax:** Многомерное расширение сигмоиды. Переводит вектор логитов в вектор распределения вероятностей для множества классов (сумма элементов равна $1$). 
  $$P_i = \frac{e^{z_i}}{\sum e^{z_j}}$$

### 4) Частные производные
Производная функции многих переменных $f(x_1, x_2, \dots, x_n)$, вычисляемая по одной конкретной переменной $x_i$. При этом все остальные переменные рассматриваются как фиксированные константы. Обозначается как $\frac{\partial f}{\partial x_i}$.

### 5) Определение градиента
**Градиент** — это вектор, составленный из всех частных производных функции многих переменных:
$$\nabla f = \left( \frac{\partial f}{\partial x_1}, \frac{\partial f}{\partial x_2}, \dots, \frac{\partial f}{\partial x_n} \right)$$
**Критическое свойство:** Градиент всегда указывает в сторону наискорейшего **роста** функции, а его длина равна скорости этого роста.

### 6) Градиентный спуск
Итерационный алгоритм оптимизации для поиска минимума функции потерь (Loss). Для движения вниз мы вычитаем вектор градиента, умноженный на шаг обучения (learning rate, $\eta$):
$$w^{(t+1)} = w^{(t)} - \eta \nabla L(w^{(t)})$$

### 7) Производная через прямую и центральную разность
Численные методы аппроксимации производных на компьютере без аналитического вывода формул:
* **Прямая разность (Forward difference):** $f'(x) \approx \frac{f(x+h) - f(x)}{h}$ (погрешность порядка $O(h)$).
* **Центральная разность (Central difference):** $f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}$ (более точная, погрешность порядка $O(h^2)$).

### 8) Линии уровня и градиент (контурный график через matplotlib)
* **Линия уровня (изолиния):** Линия на плоскости признаков, во всех точках которой функция принимает одно и то же значение $f(x, y) = \text{const}$.
* **Важное свойство:** Вектор градиента в любой точке пространства всегда строго **перпендикулярен** линии уровня (изолинии), проходящей через эту точку. В `matplotlib` визуализируется с помощью `plt.contour()`.

### 9) Momentum Gradient Descent
Модификация градиентного спуска, добавляющая «инерцию» (накапливание скорости из прошлых шагов). Помогает быстрее проходить пологие участки (плато) и преодолевать локальные минимумы или седловые точки:
$$v^{(t+1)} = \beta v^{(t)} + \eta \nabla L(w^{(t)})$$
$$w^{(t+1)} = w^{(t)} - v^{(t+1)}$$
Где $\beta \in [0, 1)$ — коэффициент удержания импульса.

## Урок №5. Цепное правило и backpropagation

### 1) Цепное правило (Chain Rule)
Правило дифференцирования сложной функции (функции от функции). Если $y = f(u)$, а $u = g(x)$, то производная $y$ по $x$ равна произведению их производных:
$$\frac{\partial y}{\partial x} = \frac{\partial y}{\partial u} \cdot \frac{\partial u}{\partial x}$$

### 2) Вычислительный граф
Представление математических вычислений в виде направленного ациклического графа. Узлы (ноды) — это математические операции ($+$, $\times$, $\sin$, $\exp$), а ребра — переменные и промежуточные тензоры. Позволяет декомпозировать сложнейшие формулы на модульные шаги.

### 3) Backpropagation (Обратное распространение ошибки)
Алгоритм автоматического вычисления градиентов в нейросетях на базе вычислительного графа:
1. **Forward pass (Прямой ход):** Расчет предсказания и значения ошибки от входа к выходу сети.
2. **Backward pass (Обратный ход):** Движение от выхода (Loss) обратно к входам с последовательным применением цепного правила (Chain Rule) на каждом узле для вычисления производных по весам.

### 4) Vanishing Gradient and Exploding Gradient
* **Затухание градиента (Vanishing Gradient):** В глубоких сетях при использовании функций вроде Sigmoid производные (которые $< 0.25$) многократно перемножаются. Градиент на первых слоях стремится к нулю, и веса перестают обучаться. Решение: использование ReLU, BatchNorm, Residual connections (ResNet).
* **Взрыв градиента (Exploding Gradient):** Обратная ситуация — при перемножении производных $> 1$ градиент экспоненциально растет, вызывая переполнение памяти (`NaN`) и полную дестабилизацию сети. Решение: обрезка градиентов (Gradient Clipping).

## Урок №6. Матричное дифференцирование

### 1) Типы производных в ML
Классификация по форме входа и выхода:
* **Скаляр по вектору:** На выходе вектор частных производных (это классический вектор-градиент $\nabla_x f$ размера $n \times 1$).
* **Вектор по вектору:** На выходе матрица Якобиана ($J$ размера $m \times n$), где строки соответствуют элементам вектора-функции, а столбцы — элементам вектора-аргумента: $J_{ij} = \frac{\partial f_i}{\partial x_j}$.
* **Скаляр по матрице:** На выходе матрица такого же размера, как исходная, где каждая ячейка содержит частную производную скаляра по конкретному весу $W_{ij}$.

### 2) Метод дифференциалов
Элегантный способ поиска сложных матричных производных без поэлементного вычисления. Базируется на поиске первого полного дифференциала скалярной функции $df$. Как только выражение приводится к каноническому виду $df = \text{Tr}(A^T dX)$, матрица $A$ объявляется искомой производной $\frac{\partial f}{\partial X}$.

### 3) Основные свойства дифференциалов
* Линейность: $d(A + B) = dA + dB$, $d(\alpha A) = \alpha dA$
* Произведение: $d(AB) = (dA)B + A(dB)$ *(строго соблюдаем порядок матриц!)*
* Транспонирование и след: $d(A^T) = (dA)^T$, $d(\text{Tr}(A)) = \text{Tr}(dA)$
* Обратная матрица: $d(A^{-1}) = -A^{-1} (dA) A^{-1}$

### 4) Канонические формы матричного дифференцирования
Связь дифференциала скаляра $df$ с производными по вектору или матрице:
1. $df = A dx \implies \frac{\partial f}{\partial x} = A^T$ (в стандартном соглашении векторов-столбцов).
2. $df = \text{Tr}(A dX) \implies \frac{\partial f}{\partial X} = A^T$

### 5) Якобиан, Гессиан
* **Якобиан ($J$):** Матрица первых частных производных вектор-функции по вектор-аргументу.
* **Гессиан ($H$):** Квадратная матрица вторых частных производных скалярной функции: $H_{ij} = \frac{\partial^2 f}{\partial x_i \partial x_j}$. Описывает кривизну многомерной поверхности локально. В выпуклых функциях Гессиан всегда полуположительно определен.

### 6) Метод Ньютона (оптимизация второго порядка)
Алгоритм поиска экстремума, использующий не только направление (градиент), но и кривизну поверхности (Гессиан). Сходится значительно быстрее градиентного спуска (квадратичная сходимость), но требует дорогого обращения матрицы Гессиана на каждом шаге:
$$w^{(t+1)} = w^{(t)} - H^{-1} \nabla f(w^{(t)})$$

### 7) Ridge-регрессия (в матричном виде)
Функция потерь с учетом $L_2$-регуляризации запишется как: $L(w) = \|Xw - y\|_2^2 + \lambda \|w\|_2^2$.
Дифференцируя её по вектору весов $w$ по методу матричного дифференцирования и приравнивая к нулю, получаем точную аналитическую формулу решения:
$$\frac{\partial L}{\partial w} = 2X^T(Xw - y) + 2\lambda w = 0 \implies w = (X^T X + \lambda I)^{-1} X^T y$$

### 8) Полуположительно определенная матрица Грама
Матрица Грама $G = X^T X$ состоит из скалярных произведений векторов. Она всегда полуположительно определена ($G \succeq 0$), поскольку для любого вектора $u$ квадратичная форма сводится к квадрату евклидовой длины нового вектора $Xu$:
$$u^T G u = u^T X^T X u = (Xu)^T (Xu) = \|Xu\|_2^2 \ge 0$$
Это свойство гарантирует, что целевая функция МНК является строго выпуклой и имеет единственный глобальный минимум.
